# Beginner Exercises (5)

**y** as the quantitative (numeric) value

**x** as the category label.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

data = {
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'revenue': [12000, 15000, 13500, 17000, 16000, 19000],
    'expenses': [8000, 9000, 8500, 10000, 9500, 11000],
    'orders': [320, 410, 375, 480, 450, 530]
}
df = pd.DataFrame(data)
df
```

```python
df.dtypes
```

Line chart — plot revenue over month. Add a title 'Monthly Revenue', x-label 'Month', y-label 'Revenue (MAD)'.

```python
df.plot(x='month', y='revenue', kind='line')
plt.title('Monthly Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue (MAD)')
plt.show()
```

Bar chart — plot orders per month. Add title and labels.

```python
df.plot(x='month', y='orders', kind='bar')
plt.title('Monthly Orders')
plt.xlabel('Month')
plt.ylabel('Orders')
plt.show()
```

Two lines on one chart — plot both revenue and expenses on the same chart. Add a legend.

```python
df.plot(x='month', y=['revenue', 'expenses'], kind='line')
plt.title("Monthy Revenue & Expenses")
plt.xlabel("Month")
plt.ylabel("Amount (MAD)")
plt.legend()
plt.show()
```

Histogram — generate 200 random ages between 20 and 60:

Plot their distribution with 10 bins. Add title 'Age Distribution'.

```python
ages = np.random.randint(20, 60, 200)

plt.hist(ages, bins=10, range=(20, 60), color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.show()
```

Pie chart — plot revenue split across first 3 months only (Jan, Feb, Mar). Add labels and percentage display.

```python
df.set_index('month')['revenue'].head(3).plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Revenue Split — Q1')
plt.ylabel('')  # hides the ugly 'revenue' ylabel pie charts get by default
plt.show()
```

# Intermediate Exercises (3)

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

data = {
    'month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'revenue': [12000, 15000, 13500, 17000, 16000, 19000],
    'expenses': [8000, 9000, 8500, 10000, 9500, 11000],
    'orders': [320, 410, 375, 480, 450, 530],
    'profit': [4000, 6000, 5000, 7000, 6500, 8000]
}
df = pd.DataFrame(data)
df
```

**1. Subplots grid**

Create a 2×2 subplot grid showing:

Top left: revenue line chart

Top right: orders bar chart

Bottom left: profit line chart

Bottom right: histogram of revenue (10 bins)

Add a title to each subplot. Use figsize=(12, 8) and plt.tight_layout().

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
x=df['month']
y1=df['revenue']
y2=df['orders']
y3=df['profit']

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Revenue')

axes[0, 1].bar(x, y2)
axes[0, 1].set_title('Orders')

axes[1, 0].plot(x, y3)
axes[1, 0].set_title('Profit')

axes[1, 1].hist(y1, bins=10)
axes[1, 1].set_title('Revenue')

plt.tight_layout()
plt.show()
```

2. Scatter plot

Plot a scatter chart of orders (x-axis) vs revenue (y-axis) to show the relationship between them. Add title, labels, and a grid.

```python
fig, ax = plt.subplots()
x=df['orders']
y=df['revenue']
ax.scatter(x, y, color='purple')
ax.set_title('Scatter Plot')
ax.set_xlabel('Orders')
ax.set_ylabel('Revenue')
ax.grid(True, linestyle='dotted', alpha=0.5, color='blue')
plt.show()
```

**3. The "client dashboard" scenario**


A logistics client wants a single figure with 2 side-by-side charts:


Left: bar chart of revenue per month

Right: pie chart of expenses per month (with percentages)


Use fig, axes = plt.subplots(1, 2, figsize=(14, 6)), add a main title with fig.suptitle('Monthly Dashboard'), and plt.tight_layout().

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
x=df['month']
y1=df['revenue']
y2=df['expenses']

# Main title
fig.suptitle('Monthly Dashboard')

axes[0].bar(x,y1)
axes[0].set_title('Revenue')

# axes[0] needs x-axis labels rotated — month names overlap on bar chart
axes[0].tick_params(axis='x', rotation=45)

axes[1].pie(y2, labels=x, autopct='%1.1f%%', startangle=90)
axes[1].set_title('Expenses')

plt.tight_layout()
plt.show()
```
