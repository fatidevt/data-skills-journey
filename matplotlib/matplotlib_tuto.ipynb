# Basics of Matplotlib

```python
### customization
# marker :
# markersize (ms)
# markerfa cecolor (mfc)
# markeredgecolor (mec)
# linestyle : solid - dashed - dotted - dashdot - None (no line)
# linewidth : thikness of the line
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,20,17,30])
y2 = np.array([11,15,27,47])
y3 = np.array([17,20,27,35])

plt.title('My Custom Title',
          fontname='serif',
          fontsize=16,
          fontweight='bold',
          color='darkblue')

plt.xlabel("Year", fontsize=20, color='cyan')
plt.ylabel("Sales", fontsize=20, color='cyan')


line_style = dict(
         marker=".",
         markersize=15,
         markerfacecolor='#9c029c',
         markeredgecolor = '#00056e',
         linestyle = 'solid',
         linewidth = 3)


plt.plot(x,y1, **line_style, color = "#b36122" )
plt.plot(x,y2, **line_style, color = "#076307" )
plt.plot(x,y3, **line_style, color = "#0517a1" )

#plt.xticks(x)

# axis = x or axis = y or axis = both
plt.tick_params(axis="x",
                colors = "cyan")

plt.show()
```

```python
# grid() = create grid lines, Help make plots easier to read by adding reference lines.
x= [1, 2, 3, 4, 5]
y= [5, 10, 15, 20, 25]

plt.grid(axis = "y",
         linewidth=2,
         color = "lightgray",
         linestyle="dashed")
plt.plot(x, y)
plt.show()
```

```python
# Bar chart = compare categories of data by representing each category with a bar
categories = np.array(["Grains", "Fruit", "Vegetables", "Proteine", "Dairy", "Sweets"])
values = [4, 3, 2, 5, 3, 1]

plt.bar(categories, values, color = 'skyblue')
#plt.barh(categories, values, color = 'skyblue')


plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

```python
# Pie chart = Circular chart divided into slices to hsow percentages of the total.
#             Good for visualizing disribution among categories

categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = np.array([300,250,275,225])
colors = ["#FFB6A6", "#FFEBD3", "#9BCEC1", "#67A2C5"]

plt.pie(values, labels=categories,
                autopct= "%1.1f%%",
                colors = colors,
                explode = [0, 0, 0, 0.1],
                shadow = True,
                startangle = 90 )


plt.title("Student Distribution")
plt.show()
```

```python
# Scatter graph = Shows the relationship btw 2 var
#                 Helps to identify a correlation (+, -, None)
#                 Example: Study hours vs Test scores

x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8 ]) # hours
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) # grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # hours
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) # grades

# alpha : transparency
# s : size of the dots
plt.scatter(x1, y1, color="skyblue",
                  alpha = 0.5,
                  s = 150,
                  label = "Class A")

plt.scatter(x2, y2, color="pink",
                  alpha = 0.5,
                  s = 150,
                  label = "Class B")

plt.title("Test Scores")
plt.xlabel("Hours Studies")
plt.ylabel("Grade")

plt.legend()
plt.show()
```

```python
# Histogram = A visual representation of the distribution of quantitative data.
#             They group values into bins (intervals)
#             and counts how many fall in each range

scores = np.random.normal(
    loc=80,     # Mean ("Center"): The average score. Most data points will cluster around 80.
    scale=10,    # Standard Deviation ("Spread"): How spread out the scores are. ~68% of scores will fall between 70 and 90 (80 ± 10).
    size=100      # Sample Size ("Amount"): Generates 100 random score values in total.
)
scores = np.clip(scores, 0, 100)
# now there is no scores bellow 0 and no scores above 100
plt.hist(scores, bins=20, color = 'lightgreen', edgecolor = 'black')
plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("nb of students")

plt.show()
```

```python
# Figure = The entire canvas
# Ax = A sigle plot (subplot)

x = np.array([1,2,3,4,5])

figures, axes = plt.subplots(2,2)
#figures, axes = plt.subplots(3,2)  #(r,c)

axes[0, 0].plot(x, x*2, color="red")
axes[0, 0].set_title('x*2')

axes[0, 1].plot(x, x**2, color="blue")
axes[0, 1].set_title('x**2')

axes[1, 0].plot(x, x**3, color="green")
axes[1, 0].set_title('x**3')

axes[1, 1].plot(x, x**4, color="purple")
axes[1, 1].set_title('x**4')

plt.tight_layout()
plt.show()
```
